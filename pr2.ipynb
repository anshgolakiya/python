```python
help(list)
```

```python
x = [10,30,20]
print("data type of x is",type(x))
print("list x is",x)
x.append(40)
print("list x after using append method is",x)
y = x.copy()
print("list y after using copy method is",y)
print("list x after using count method is",x.count(10))
z = [90,50,60,70]
x.extend(z)
print("list x after using extend method is(x.extend(z))",x)
print("index of 20 in list x is",x.index(20))
print("list x after using insert method is(x.insert(2,100))",x.insert(2,100))
print("list x after using pop method is",x.pop())
print("list x after using reverse method is",x.reverse())
print("list x after using sort method is",x.sort())
print("list x after using clear method is",x.clear())
```

```python
help(set)
```

```python
s= {10,30,20}
print("data type of set is",type(s))
print("set is",s) # output is {10,20,30} because set is unordered collection of data
s.add(40)
print("set after using add method is",s)
s1 = s.copy()
print("set s1 after using copy method is",s1)
s2 = {90,50,20,10,70}
print("set after using difference method is",s.difference(s2))
print("set after using discard method is",s.discard(20))
print("set after using intersection method is",s.intersection(s2))
print("set after using isdisjoint method is",s.isdisjoint(s2))
print("set after using issubset method is",s.issubset(s2))
print("set after using issuperset method is",s.issuperset(s2))
print("set after using pop method is",s.pop())
print("set after using remove method is",s.remove(30))
print("set after using clear method is",s.clear())
```

```python
d = {"name":"Ansh","age":20,"city":"surat"}
print("data type of dictionary is",type(d))
print("dictionary is",d)
print("dictionary after using get method) is",d.get("name"))
print("dictionary after using items method is",d.items())
print("dictionary after using keys method is",d.keys())
print("dictionary after using values method is",d.values())
print("dictionary after using pop method is",d.pop("age"))
d.update({"surname" : "Golakiya"})
print("dictionaty after using update method is",d)
```
