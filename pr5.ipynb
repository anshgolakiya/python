```python
import csv
with open("demo.csv",'w') as file :
  writer = csv.writer(file)
  writer.writerow(["ID","Name","marks"])
  writer.writerow([101,"Ansh",90])
  writer.writerow([102,"Raj",80])
```

```python
import csv
with open("demo.csv",'r') as file :
  reader = csv.reader(file)
  for row in reader:
    print(row)
```
